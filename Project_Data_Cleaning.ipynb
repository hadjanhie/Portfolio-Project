# Data Cleaning in Pandas

```python
import pandas as pd
```

```python
df=pd.read_excel(r"C:\Users\choul\Downloads\Bootcamp tutorials\Python\Customer Call List.xlsx")
df
```

```python
df=df.drop_duplicates()
df
```

```python
df= df.drop(columns='Not_Useful_Column')
df
```

```python
#df["Last_Name"]=df["Last_Name"].str.lstrip("...")
#df["Last_Name"]=df["Last_Name"].str.lstrip('/')
#df["Last_Name"]=df["Last_Name"].str.rstrip('_')
df["Last_Name"]=df["Last_Name"].str.strip("123./_")
df # there is NAN as last name, remove it
```

```python

df["Phone_Number"]= df["Phone_Number"].apply(lambda x:str(x))
df
```

```python
df["Phone_Number"]= df["Phone_Number"].str.replace('[^a-zA-Z0-9]','',regex=True)
df
```

```python
df["Phone_Number"]=df["Phone_Number"].apply(lambda x:x[0:3]+ '-'+x[3:6]+'-'+x[6:10])
df
```

```python
df["Phone_Number"]=df["Phone_Number"].str.replace('nan--','', regex=True)
df["Phone_Number"]=df["Phone_Number"].str.replace('Na--','',regex=True)
df
```

```python
df[['Street_Address','State','Zip_Code']]=df['Address'].str.split(',',n=2,expand=True)
df 
#n=2 → max 2 splits from the left → 3 resulting columns.

```

```python
df= df.drop(columns='Address')
df
```

```python
df=df.fillna('')
df = df.replace('N/a', '', regex=True)
df
```

```python
df['Paying Customer']=df['Paying Customer'].str.replace('Yes','Y')

df['Paying Customer']=df['Paying Customer'].str.replace('No','N')
df['Paying Customer']=df['Paying Customer'].str.replace('Y','Yes')
df['Paying Customer']=df['Paying Customer'].str.replace('N','No')
df
```

```python
df['Do_Not_Contact']=df['Do_Not_Contact'].str.replace('Yes','Y')

df['Do_Not_Contact']=df['Do_Not_Contact'].str.replace('No','N')
df['Do_Not_Contact']=df['Do_Not_Contact'].str.replace('Y','Yes')
df['Do_Not_Contact']=df['Do_Not_Contact'].str.replace('N','No')
df
```

```python
for x in df.index:
    if df.loc[x,"Do_Not_Contact"]=='Yes':
        df.drop(x,inplace=True)

df               
```

```python
for x in df.index:
    if df.loc[x,"Phone_Number"]=='':
        df.drop(x,inplace=True)

df
# Another way
#df=df.dropna(subset="Phone_Number"),inplace=True)
```

```python
df=df.reset_index(drop=True)

#print(df[df['Do_Not_Contact'] == ''])   # Check for empty strings
# Replace empty strings with 'No'
df['Do_Not_Contact'] = df['Do_Not_Contact'].replace('', 'No')


df
```
